# Fabric Defect Classification — 04. MobileNetV2 Fine-Tuning (Transfer Learning)

This notebook adds a **transfer-learning** model to the comparison — a MobileNetV2 pretrained on ImageNet, then fine-tuned on the fabric-defect dataset. It serves as a strong "ceiling" reference against the three from-scratch CNNs (Baseline, Improved, Tuned).

## Why MobileNetV2
* Pretrained on ImageNet (1.28M images), so the early layers already encode useful texture, edge, and colour-gradient features — exactly the low-level structures that distinguish fabric defects.
* Small (~2.2M parameters in the feature backbone), so fine-tuning runs comfortably on CPU and is fast on any GPU.
* Uses inverted-residual blocks with depthwise-separable convolutions — efficient and well-suited to small datasets.

## Two-stage training strategy
1. **Stage 1 — Feature extraction (5 epochs).** Freeze the entire MobileNetV2 backbone and train only the new classifier head. This lets the head learn to map ImageNet features onto our four classes without disturbing the pretrained weights.
2. **Stage 2 — Fine-tuning (15 epochs).** Unfreeze the last block of the backbone and continue training with a **much smaller** learning rate (10× lower). This adapts the high-level features to fabric textures without destroying what was learned on ImageNet.

## Data-leakage controls
Same as the other model notebooks — reads the fixed `fabric_split/` produced by `00_Data_Preparation_and_EDA.ipynb`, applies augmentation only to the training split, computes class weights from training labels only, and touches the test split exactly once at the end.

## A note on input size
MobileNetV2 was pretrained on 224×224 images. We resize to **224×224** here (not 128×128 like the other notebooks) because using the pretrained model at its native resolution gives noticeably better features. This is one place transfer learning constrains us — but it's a worthwhile constraint.

## 1. Imports & setup

```python
import os
import copy
import json
import random
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.utils.data import DataLoader
from torchvision import datasets, transforms, models
from sklearn.metrics import (
    confusion_matrix, classification_report, precision_recall_fscore_support, roc_curve, auc
)

SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else
                      'mps'  if torch.backends.mps.is_available() else 'cpu')
print(f'Using device: {device}')

sns.set_style('whitegrid')
plt.rcParams['figure.dpi'] = 110
```

## 2. Dataset — 224×224 resize for MobileNetV2

Same `fabric_split/` directory as the other notebooks; only the resize dimensions change. Augmentation is again train-only.

```python
DATA_DIR = 'fabric_split'
IMG_SIZE = 224   # MobileNetV2's native input size
BATCH    = 32

# ImageNet statistics — these are the values MobileNetV2 was pretrained with,
# so we must normalise the same way at inference time.
MEAN = [0.485, 0.456, 0.406]
STD  = [0.229, 0.224, 0.225]

data_transforms = {
    'train': transforms.Compose([
        transforms.Resize((IMG_SIZE, IMG_SIZE)),
        transforms.RandomHorizontalFlip(),
        transforms.RandomRotation(15),
        transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.1),
        transforms.ToTensor(),
        transforms.Normalize(MEAN, STD),
    ]),
    'val': transforms.Compose([
        transforms.Resize((IMG_SIZE, IMG_SIZE)),
        transforms.ToTensor(),
        transforms.Normalize(MEAN, STD),
    ]),
    'test': transforms.Compose([
        transforms.Resize((IMG_SIZE, IMG_SIZE)),
        transforms.ToTensor(),
        transforms.Normalize(MEAN, STD),
    ]),
}

image_datasets = {
    s: datasets.ImageFolder(os.path.join(DATA_DIR, s), data_transforms[s])
    for s in ['train', 'val', 'test']
}
dataloaders = {
    'train': DataLoader(image_datasets['train'], batch_size=BATCH, shuffle=True,  num_workers=0),
    'val':   DataLoader(image_datasets['val'],   batch_size=BATCH, shuffle=False, num_workers=0),
    'test':  DataLoader(image_datasets['test'],  batch_size=BATCH, shuffle=False, num_workers=0),
}
dataset_sizes = {s: len(image_datasets[s]) for s in ['train', 'val', 'test']}
class_names   = image_datasets['train'].classes
num_classes   = len(class_names)

print('Classes      :', class_names)
print('Dataset sizes:', dataset_sizes)
```

## 3. Class-imbalance weights (from training set only)

```python
train_labels = [label for _, label in image_datasets['train'].samples]
class_counts = Counter(train_labels)
total        = sum(class_counts.values())
class_weights = [total / (num_classes * class_counts[i]) for i in range(num_classes)]
class_weights_tensor = torch.FloatTensor(class_weights).to(device)
print('Class weights:', dict(zip(class_names, [round(w, 4) for w in class_weights])))
```

## 4. Load pretrained MobileNetV2 & replace the classifier

We swap out the final 1000-way ImageNet classifier for a small head matched to our 4 fabric defect classes. The backbone (`model.features`) is kept untouched for Stage 1.

```python
# Load MobileNetV2 with ImageNet weights
model = models.mobilenet_v2(weights=models.MobileNet_V2_Weights.IMAGENET1K_V1)

# Replace the ImageNet head (Linear(1280 -> 1000)) with our own
# MobileNetV2's classifier is: Sequential(Dropout(0.2), Linear(1280, 1000))
in_features = model.classifier[1].in_features  # 1280
model.classifier = nn.Sequential(
    nn.Dropout(0.3),
    nn.Linear(in_features, 256),
    nn.ReLU(inplace=True),
    nn.Dropout(0.3),
    nn.Linear(256, num_classes),
)
model = model.to(device)

total_params     = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f'Total parameters    : {total_params:,}')
print(f'Trainable parameters: {trainable_params:,}')
print(f'\nNew classifier head:\n{model.classifier}')
```

## 5. Helper: training loop

One function used for both stages — the only thing that changes between stages is **which parameters require gradients** and the **learning rate**.

```python
def run_training(model, optimizer, scheduler, criterion, num_epochs, stage_name='train', history=None):
    """Train `model` for `num_epochs`, return updated history and best (val_acc, weights)."""
    if history is None:
        history = {'train_loss': [], 'val_loss': [], 'train_acc': [], 'val_acc': [], 'lr': [], 'stage': []}
    best_acc = max(history['val_acc']) if history['val_acc'] else 0.0
    best_wts = copy.deepcopy(model.state_dict())

    for epoch in range(num_epochs):
        print(f'[{stage_name}] Epoch {epoch+1:02d}/{num_epochs}  (lr = {optimizer.param_groups[0]["lr"]:.6f})')
        for phase in ['train', 'val']:
            model.train() if phase == 'train' else model.eval()
            running_loss, running_corrects = 0.0, 0
            for inputs, labels in dataloaders[phase]:
                inputs, labels = inputs.to(device), labels.to(device)
                optimizer.zero_grad()
                with torch.set_grad_enabled(phase == 'train'):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)
                    if phase == 'train':
                        loss.backward()
                        optimizer.step()
                running_loss     += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)
            epoch_loss = running_loss / dataset_sizes[phase]
            epoch_acc  = (running_corrects.float() / dataset_sizes[phase]).item()
            history[f'{phase}_loss'].append(epoch_loss)
            history[f'{phase}_acc'].append(epoch_acc)
            print(f'  {phase:<5s} loss={epoch_loss:.4f}  acc={epoch_acc:.4f}')
            if phase == 'val' and epoch_acc > best_acc:
                best_acc = epoch_acc
                best_wts = copy.deepcopy(model.state_dict())
        history['lr'].append(optimizer.param_groups[0]['lr'])
        history['stage'].append(stage_name)
        if scheduler is not None:
            scheduler.step()
    return history, best_acc, best_wts
```

## 6. Stage 1 — Feature extraction (backbone frozen)

Freeze every parameter in `model.features` so only the classifier head trains. This is the safe first move when fine-tuning: it adapts the output mapping without risking the pretrained features.

```python
for param in model.features.parameters():
    param.requires_grad = False

stage1_trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
stage1_frozen    = sum(p.numel() for p in model.parameters() if not p.requires_grad)
print(f'Stage 1 trainable parameters: {stage1_trainable:,}')
print(f'Stage 1 frozen parameters   : {stage1_frozen:,}')

criterion = nn.CrossEntropyLoss(weight=class_weights_tensor)
STAGE1_EPOCHS = 5
STAGE1_LR     = 1e-3

optimizer = optim.Adam(
    [p for p in model.parameters() if p.requires_grad],
    lr=STAGE1_LR,
    weight_decay=1e-4,
)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=STAGE1_EPOCHS)

history, stage1_best_acc, stage1_best_wts = run_training(
    model, optimizer, scheduler, criterion, STAGE1_EPOCHS, stage_name='stage1_head'
)
print(f'\nStage 1 complete. Best val acc = {stage1_best_acc:.4f}')
model.load_state_dict(stage1_best_wts)
```

## 7. Stage 2 — Fine-tuning the last block of MobileNetV2

Now we unfreeze the **last inverted-residual block** of the backbone (`features[14:]`) and continue training. The learning rate is dropped by 10× — high learning rates here would erase the carefully-learned ImageNet features.

Unfreezing only the last block (rather than the whole backbone) is a common transfer-learning compromise: it gives the model freedom to adapt high-level features to fabric textures, while keeping the low-level edge / colour / texture filters from ImageNet intact.

```python
# Unfreeze the last few feature blocks. MobileNetV2 has 19 blocks (features[0]..features[18]);
# the last block (features[18]) is the 1×1 conv that produces the 1280-channel feature map,
# and features[14:18] are the final inverted-residual blocks.
for name, param in model.features.named_parameters():
    # Block index is the first token of `name` for sub-modules in nn.Sequential
    block_idx = int(name.split('.')[0])
    if block_idx >= 14:
        param.requires_grad = True

stage2_trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
stage2_frozen    = sum(p.numel() for p in model.parameters() if not p.requires_grad)
print(f'Stage 2 trainable parameters: {stage2_trainable:,}')
print(f'Stage 2 frozen parameters   : {stage2_frozen:,}')

STAGE2_EPOCHS = 15
STAGE2_LR     = 1e-4   # 10× lower than Stage 1

optimizer = optim.Adam(
    [p for p in model.parameters() if p.requires_grad],
    lr=STAGE2_LR,
    weight_decay=1e-4,
)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=STAGE2_EPOCHS)

history, best_acc, best_wts = run_training(
    model, optimizer, scheduler, criterion, STAGE2_EPOCHS,
    stage_name='stage2_finetune', history=history
)
print(f'\nFine-tuning complete. Overall best val acc = {best_acc:.4f}')
model.load_state_dict(best_wts)
torch.save(best_wts, 'mobilenet_finetuned_weights.pth')
with open('mobilenet_history.json', 'w') as f:
    json.dump(history, f, indent=2)
```

## 8. Training-curve visualisations

The vertical dashed line marks the transition from Stage 1 (head-only) to Stage 2 (fine-tuning). Watch what happens to val accuracy when the backbone unfreezes — typically a noticeable jump.

```python
split_at = STAGE1_EPOCHS  # epoch index where stage 2 begins
epochs   = np.arange(len(history['train_loss']))

fig, axes = plt.subplots(1, 3, figsize=(16, 4.2))

axes[0].plot(epochs, history['train_loss'], '-o', label='Train', linewidth=2, markersize=4)
axes[0].plot(epochs, history['val_loss'],   '-s', label='Val',   linewidth=2, markersize=4)
axes[0].axvline(split_at - 0.5, color='black', linestyle='--', alpha=0.5)
axes[0].text(split_at - 0.6, axes[0].get_ylim()[1] * 0.95, 'Stage 2\nfine-tune', ha='right', fontsize=9, alpha=0.7)
axes[0].set_title('Loss', fontweight='bold')
axes[0].set_xlabel('Epoch'); axes[0].set_ylabel('Cross-entropy')
axes[0].legend(); axes[0].grid(alpha=0.3)

axes[1].plot(epochs, history['train_acc'], '-o', label='Train', linewidth=2, markersize=4)
axes[1].plot(epochs, history['val_acc'],   '-s', label='Val',   linewidth=2, markersize=4)
axes[1].axvline(split_at - 0.5, color='black', linestyle='--', alpha=0.5)
axes[1].axhline(best_acc, color='red', linestyle='--', alpha=0.5, label=f'Best val = {best_acc:.3f}')
axes[1].set_title('Accuracy', fontweight='bold')
axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('Accuracy')
axes[1].legend(); axes[1].grid(alpha=0.3)

axes[2].plot(epochs, history['lr'], '-o', color='purple', linewidth=2, markersize=4)
axes[2].axvline(split_at - 0.5, color='black', linestyle='--', alpha=0.5)
axes[2].set_yscale('log')
axes[2].set_title('Learning rate (log scale)', fontweight='bold')
axes[2].set_xlabel('Epoch'); axes[2].set_ylabel('LR')
axes[2].grid(alpha=0.3, which='both')

plt.suptitle('MobileNetV2 fine-tuning — two-stage training', fontweight='bold')
plt.tight_layout()
plt.savefig('mobilenet_training_curves.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Overlay against all earlier models (if available)
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
model_curves = []
for fname, label, style in [
    ('baseline_history.json', 'Baseline', '--o'),
    ('improved_history.json', 'Improved', '-.s'),
]:
    if os.path.exists(fname):
        with open(fname) as f:
            h = json.load(f)
        model_curves.append((label, h, style))

for label, h, style in model_curves:
    axes[0].plot(h['val_loss'], style, label=f'{label} val loss', alpha=0.7)
    axes[1].plot(h['val_acc'],  style, label=f'{label} val acc',  alpha=0.7)
axes[0].plot(history['val_loss'], '-d', label='MobileNetV2 val loss', linewidth=2, color='crimson')
axes[1].plot(history['val_acc'],  '-d', label='MobileNetV2 val acc',  linewidth=2, color='crimson')

axes[0].set_title('Val loss — all models', fontweight='bold')
axes[0].set_xlabel('Epoch'); axes[0].legend(); axes[0].grid(alpha=0.3)
axes[1].set_title('Val accuracy — all models', fontweight='bold')
axes[1].set_xlabel('Epoch'); axes[1].legend(); axes[1].grid(alpha=0.3)
plt.tight_layout()
plt.savefig('mobilenet_vs_others_curves.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 9. Test-set evaluation

```python
model.eval()
y_true, y_pred, y_prob = [], [], []
with torch.no_grad():
    for inputs, labels in dataloaders['test']:
        inputs, labels = inputs.to(device), labels.to(device)
        outputs = model(inputs)
        probs   = F.softmax(outputs, dim=1)
        _, preds = torch.max(outputs, 1)
        y_true.extend(labels.cpu().numpy())
        y_pred.extend(preds.cpu().numpy())
        y_prob.extend(probs.cpu().numpy())
y_true = np.array(y_true); y_pred = np.array(y_pred); y_prob = np.array(y_prob)

mobilenet_test_acc = (y_true == y_pred).mean()
print(f'Test accuracy: {mobilenet_test_acc:.4f}')
print('\nClassification report:')
print(classification_report(y_true, y_pred, target_names=class_names, digits=4))
```

```python
cm     = confusion_matrix(y_true, y_pred)
cm_pct = cm.astype(float) / cm.sum(axis=1, keepdims=True)
fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))
sns.heatmap(cm,     annot=True, fmt='d',   cmap='Reds',
            xticklabels=class_names, yticklabels=class_names, ax=axes[0])
axes[0].set_title('Confusion matrix — counts', fontweight='bold')
axes[0].set_xlabel('Predicted'); axes[0].set_ylabel('True')
sns.heatmap(cm_pct, annot=True, fmt='.2f', cmap='Reds',
            xticklabels=class_names, yticklabels=class_names, ax=axes[1], vmin=0, vmax=1)
axes[1].set_title('Confusion matrix — row-normalised', fontweight='bold')
axes[1].set_xlabel('Predicted'); axes[1].set_ylabel('True')
plt.tight_layout()
plt.savefig('mobilenet_confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
p, r, f1, sup = precision_recall_fscore_support(y_true, y_pred, labels=range(num_classes), zero_division=0)
metrics_df = pd.DataFrame({'class': class_names, 'precision': p, 'recall': r, 'f1': f1, 'support': sup})
print(metrics_df.round(4).to_string(index=False))

fig, ax = plt.subplots(figsize=(10, 5))
x = np.arange(num_classes); w = 0.27
ax.bar(x - w, metrics_df['precision'], w, label='Precision', color='#1f77b4')
ax.bar(x,     metrics_df['recall'],    w, label='Recall',    color='#ff7f0e')
ax.bar(x + w, metrics_df['f1'],        w, label='F1',        color='#2ca02c')
ax.set_xticks(x); ax.set_xticklabels(class_names); ax.set_ylim(0, 1.05)
ax.set_title('Per-class precision / recall / F1 — MobileNetV2 fine-tuned', fontweight='bold')
ax.set_ylabel('Score'); ax.legend(); ax.grid(axis='y', alpha=0.3)
for i in range(num_classes):
    ax.text(i - w, metrics_df['precision'][i] + 0.01, f'{metrics_df["precision"][i]:.2f}', ha='center', fontsize=8)
    ax.text(i,     metrics_df['recall'][i]    + 0.01, f'{metrics_df["recall"][i]:.2f}',    ha='center', fontsize=8)
    ax.text(i + w, metrics_df['f1'][i]        + 0.01, f'{metrics_df["f1"][i]:.2f}',        ha='center', fontsize=8)
plt.tight_layout()
plt.savefig('mobilenet_per_class_metrics.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ROC curves
fig, ax = plt.subplots(figsize=(7.5, 6))
for i, cls in enumerate(class_names):
    y_bin = (y_true == i).astype(int)
    fpr, tpr, _ = roc_curve(y_bin, y_prob[:, i])
    ax.plot(fpr, tpr, linewidth=2, label=f'{cls} (AUC = {auc(fpr, tpr):.3f})')
ax.plot([0, 1], [0, 1], 'k--', alpha=0.4)
ax.set_xlabel('False positive rate'); ax.set_ylabel('True positive rate')
ax.set_title('One-vs-rest ROC curves — MobileNetV2 fine-tuned', fontweight='bold')
ax.legend(loc='lower right'); ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig('mobilenet_roc_curves.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 10. Sample predictions

```python
def denormalise(t):
    mean = torch.tensor(MEAN).view(3, 1, 1)
    std  = torch.tensor(STD).view(3, 1, 1)
    return (t * std + mean).clamp(0, 1)

all_imgs, all_true, all_pred, all_conf = [], [], [], []
model.eval()
with torch.no_grad():
    for inputs, labels in dataloaders['test']:
        inputs = inputs.to(device); labels = labels.to(device)
        probs  = F.softmax(model(inputs), dim=1)
        conf, preds = torch.max(probs, dim=1)
        all_imgs.append(inputs.cpu()); all_true.append(labels.cpu())
        all_pred.append(preds.cpu());  all_conf.append(conf.cpu())
all_imgs = torch.cat(all_imgs); all_true = torch.cat(all_true)
all_pred = torch.cat(all_pred); all_conf = torch.cat(all_conf)

correct_idx = (all_pred == all_true).nonzero(as_tuple=True)[0].tolist()
wrong_idx   = (all_pred != all_true).nonzero(as_tuple=True)[0].tolist()
rng = random.Random(SEED)
correct_sample = rng.sample(correct_idx, min(6, len(correct_idx)))
wrong_sample   = rng.sample(wrong_idx,   min(6, len(wrong_idx))) if wrong_idx else []

def plot_grid(indices, title, save):
    if not indices:
        print(f'(no examples for: {title})'); return
    n = len(indices)
    fig, axes = plt.subplots(1, n, figsize=(3 * n, 3.5))
    if n == 1: axes = [axes]
    for ax, i in zip(axes, indices):
        ax.imshow(denormalise(all_imgs[i]).permute(1, 2, 0).numpy())
        t, p, c = class_names[all_true[i]], class_names[all_pred[i]], all_conf[i].item()
        color = 'green' if t == p else 'red'
        ax.set_title(f'T: {t}\nP: {p} ({c*100:.0f}%)', color=color, fontsize=9)
        ax.axis('off')
    plt.suptitle(title, fontweight='bold')
    plt.tight_layout()
    plt.savefig(save, dpi=150, bbox_inches='tight')
    plt.show()

plot_grid(correct_sample, 'Correct predictions — MobileNetV2',  'mobilenet_correct.png')
plot_grid(wrong_sample,   'Incorrect predictions — MobileNetV2','mobilenet_wrong.png')
```

## 11. Save metrics & final four-way comparison

```python
p_macro, r_macro, f1_macro, _ = precision_recall_fscore_support(y_true, y_pred, average='macro', zero_division=0)
mobilenet_metrics = {
    'model_name'    : 'MobileNetV2 (fine-tuned)',
    'best_val_acc'  : float(best_acc),
    'test_acc'      : float(mobilenet_test_acc),
    'test_precision_macro': float(p_macro),
    'test_recall_macro'   : float(r_macro),
    'test_f1_macro'       : float(f1_macro),
    'num_params_total'    : int(total_params),
    'num_params_trainable_final': int(stage2_trainable),
    'stage1_epochs' : STAGE1_EPOCHS,
    'stage2_epochs' : STAGE2_EPOCHS,
    'class_names'   : class_names,
}
with open('mobilenet_metrics.json', 'w') as f:
    json.dump(mobilenet_metrics, f, indent=2)
print('Saved mobilenet_metrics.json')
print(json.dumps(mobilenet_metrics, indent=2))
```

```python
# Build the four-way comparison table
rows = []
for fname, label in [('baseline_metrics.json',  'Baseline CNN'),
                      ('improved_metrics.json',  'Improved CNN'),
                      ('tuned_metrics.json',     'Tuned CNN'),
                      ('mobilenet_metrics.json', 'MobileNetV2 (FT)')]:
    if os.path.exists(fname):
        with open(fname) as f:
            m = json.load(f)
        rows.append({
            'Model'    : label,
            'Val Acc'  : m.get('best_val_acc'),
            'Test Acc' : m.get('test_acc'),
            'Test Prec (macro)' : m.get('test_precision_macro'),
            'Test Rec (macro)'  : m.get('test_recall_macro'),
            'Test F1 (macro)'   : m.get('test_f1_macro'),
        })
    else:
        print(f'(skipped: {fname} not found)')

comp_df = pd.DataFrame(rows)
print('\nFinal four-way comparison:')
print(comp_df.round(4).to_string(index=False))
comp_df.to_csv('final_comparison_with_mobilenet.csv', index=False)
```

```python
if len(comp_df) >= 2:
    metrics_to_plot = ['Val Acc', 'Test Acc', 'Test Prec (macro)', 'Test Rec (macro)', 'Test F1 (macro)']
    fig, ax = plt.subplots(figsize=(14, 5.5))
    x = np.arange(len(metrics_to_plot)); w = 0.8 / len(comp_df)
    colors = ['#999999', '#2ca02c', '#9467bd', '#d62728']
    for i, (_, row) in enumerate(comp_df.iterrows()):
        vals = [row[m] for m in metrics_to_plot]
        bars = ax.bar(x + i*w - w*(len(comp_df)-1)/2, vals, w, label=row['Model'], color=colors[i % 4])
        for b, v in zip(bars, vals):
            ax.text(b.get_x() + b.get_width()/2, v + 0.005, f'{v:.3f}', ha='center', fontsize=7.5)
    ax.set_xticks(x); ax.set_xticklabels(metrics_to_plot)
    ax.set_ylabel('Score'); ax.set_ylim(0, 1.12)
    ax.set_title('Four-way model comparison on the held-out test set', fontweight='bold')
    ax.legend(loc='lower right'); ax.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.savefig('final_comparison_four_way.png', dpi=150, bbox_inches='tight')
    plt.show()
```

```python
# Parameter-count comparison — useful for the report's "efficiency" discussion
param_rows = []
for fname, label in [('baseline_metrics.json',  'Baseline CNN'),
                      ('improved_metrics.json',  'Improved CNN'),
                      ('tuned_metrics.json',     'Tuned CNN'),
                      ('mobilenet_metrics.json', 'MobileNetV2 (FT)')]:
    if os.path.exists(fname):
        with open(fname) as f:
            m = json.load(f)
        n = m.get('num_params') or m.get('num_params_total')
        param_rows.append({'Model': label, 'Total parameters': n,
                            'Test Acc': m.get('test_acc')})
if param_rows:
    pdf = pd.DataFrame(param_rows)
    fig, ax = plt.subplots(figsize=(8.5, 5))
    ax.scatter(pdf['Total parameters'], pdf['Test Acc'], s=180,
               c=['#999999', '#2ca02c', '#9467bd', '#d62728'][:len(pdf)])
    for _, row in pdf.iterrows():
        ax.annotate(row['Model'], (row['Total parameters'], row['Test Acc']),
                    xytext=(8, 6), textcoords='offset points', fontsize=10)
    ax.set_xscale('log')
    ax.set_xlabel('Total parameters (log scale)')
    ax.set_ylabel('Test accuracy')
    ax.set_title('Accuracy vs model size', fontweight='bold')
    ax.grid(alpha=0.3, which='both')
    plt.tight_layout()
    plt.savefig('final_accuracy_vs_size.png', dpi=150, bbox_inches='tight')
    plt.show()
```

## 12. Single-image inference helper

```python
def predict_single_image(image_path: str):
    model.eval()
    try:
        image = Image.open(image_path).convert('RGB')
    except Exception as e:
        print(f'Could not load {image_path}: {e}'); return
    x = data_transforms['val'](image).unsqueeze(0).to(device)
    with torch.no_grad():
        probs = F.softmax(model(x), dim=1)[0]
        conf, idx = torch.max(probs, 0)
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))
    ax1.imshow(image); ax1.axis('off')
    ax1.set_title(f'Predicted: {class_names[idx]} ({conf*100:.1f}%)')
    ax2.barh(class_names, probs.cpu().numpy(), color='crimson')
    ax2.set_xlim(0, 1); ax2.set_xlabel('Probability')
    ax2.set_title('Class probabilities')
    plt.tight_layout(); plt.show()

# Example usage:
# predict_single_image('fabric_split/test/hole/some_image.jpg')
```
